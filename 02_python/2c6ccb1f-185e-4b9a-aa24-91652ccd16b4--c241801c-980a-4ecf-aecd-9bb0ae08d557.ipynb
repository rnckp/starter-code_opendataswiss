## Description

La loi fédérale sur l&#x27;aménagement du territoire &#40;LAT&#41; oblige les cantons à désigner les parties du territoire qui sont menacées par des forces naturelles ou par des nuisances. Les dangers représentés dans ce jeu de données, sont les dangers géologiques de glissements de terrain permanents profonds. Le danger est classifié selon les recommandations de la Confédération : danger élevé - rouge = zone d&#x27;interdiction; danger moyen - bleu = zone de réglementation; danger faible - jaune = zone de sensibilisation; aucun danger connu - blanc = zone où aucun danger n&#x27;est déterminé selon l&#x27;état des connaissances actuelles; Le GESDEC a proposé, en 2012, une transcription simplifiée de la carte de phénomènes &#40;= zones instables&#42;&#41; en carte des dangers. L&#x27;estimation du degré de danger est faite selon une matrice propre au phénomène. Dans celle-ci, les caractéristiques de chaque phénomène sont appliquées et un degré de danger est défini. Glissements de terrain permanents &#40;GLP&#41;: Selon les standards fédéraux, la caractéristique déterminant le degré de danger pour les glissements de terrain permanents est la vitesse moyenne. Un glissement se définit également par sa profondeur. Pour les glissements de terrain sur le territoire genevois, la carte des zones instables contient deux catégories de vitesses : peu actif ou lent/très lents &#40;correspondant à une vitesse &lt;5 cm/an&#41; et actif &#40;&gt;5 cm/an&#41; et deux catégories de profondeurs : superficiel &#40;correspondant à une profondeur &lt;2 m&#41; et profond &#40;&gt;2 m&#41;.
```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                # Detect file compression after HTTP compression has been decoded.
                if kwargs.get("compression", "infer") == "infer" and source.peek(2).startswith(
                    b"\x1f\x8b"
                ):
                    kwargs["compression"] = "gzip"
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
