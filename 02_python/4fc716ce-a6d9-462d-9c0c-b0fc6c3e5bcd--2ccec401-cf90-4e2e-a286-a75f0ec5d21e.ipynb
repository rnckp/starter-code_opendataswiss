## Description

This dataset contains stable isotope ratios of oxygen and hydrogen &#40;O-18 and H-2&#41; and electrical conductivity &#40;EC&#41; in overland flow &#40;OF&#41;, topsoil interflow &#40;TIF&#41;, soil water, groundwater, and rainwater for 12 rainfall events during the snow-free seasons of 2021 and 2022 in the Studibach catchment, Alptal, Switzerland. <br>Overland flow, topsoil interflow &#40;i.e., lateral flow from the more densely rooted soil layer&#41;, soil water, and groundwater were collected at 14 small trenched runoff plots &#40;1 x 3 m&#41; in three subcatchments &#40;C2, C3, C5&#41; of the Studibach catchment. Overland flow was collected from the surface up to 2-5 cm depth and thus includes biomat flow. A trench was used to collect topsoil interflow &#40;up to 60 cm below the surface; see Table 1 in the manuscript for details&#41;. Rainwater was collected at two locations in the Studibach catchment &#40;in C3 and C5, respectively&#41;. Soil water was collected in between the events from suction lysimeters installed at 12.5 cm and 20 cm below the soil surface in the middle of each plot. Groundwater was collected between events from wells installed near the plots up to the soil-bedrock interface. <br>The water samples were analysed for the stable isotopes of oxygen and hydrogen &#40;O-18 and H-2&#41; using with a cavity ring-down spectroscope &#40;CRDS; L2140-i or L2130-i, Picarro, Inc., USA&#41; at the Chair of Hydrology at the University of Freiburg, Germany. The isotope ratios are reported in per mil &#40;‰&#41; relative to Vienna Standard Mean Ocean Water.<br>A more detailed description of the field setup, data collection and preparation can be found in Leuteritz et al. &#40;in press&#41;.<br>The dataset contains in addition to the sample identifier and the site and event identifiers that are used in the publication, also the sample collection date &#40;typically one day after the event&#41; and the coordinates for each plot &#40;coordinate system: WGS84&#41;. For the plots for which samples were collected with an automatic sampler &#40;ISCO&#41;, the date and time &#40;UTC&#41; of sample collection are given as well.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
