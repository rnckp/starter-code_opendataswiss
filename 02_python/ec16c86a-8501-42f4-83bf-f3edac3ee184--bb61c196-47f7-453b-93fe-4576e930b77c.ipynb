## Description

This dataset consists of a series of full snow profiles collected at the Weissfluhjoch field site in Davos, Switzerland, over the two snow seasons of 2022-2023 and 2023-2024. This series of snow profiles was collected at high vertical and temporal resolution, with a specific focus on the melting seasons, during which up to 3 profiles per week were sampled. This dataset is part of the measurement campaign conducted within the SnowTinel project, which aims to explore the Sentinel-1 SAR backscattering response to melting Alpine snowpacks. Therefore, this dataset contains manual measurements of the main properties of snow that are responsible for scattering, i.e., temperature, density, specific surface area &#40;SSA&#41;, liquid water content &#40;LWC&#41;, surface roughness, and snow water equivalent &#40;SWE&#41;. <br>The dataset comprises 85 snow profiles: 38 carried out in 2022-2023 and 47 in 2023-2024. <br>  - Profiles of snow &#42;&#42;temperature&#42;&#42; were sampled at a vertical resolution of 10 cm &#40;2022-2023&#41; and 5 cm &#40;2023-2024&#41; using HI98501 Checktemp from Hanna Instruments;<br>   - Profiles of snow &#42;&#42;density&#42;&#42; were sampled at a vertical resolution of 3 cm using a box density cutter and a digital scale;<br>   - Profiles of snow &#42;&#42;specific surface area&#42;&#42; were sampled at a vertical resolution of 4 cm using the InfraSnow sensor from FPGA;<br>   - Profiles of snow &#42;&#42;liquid water content&#42;&#42; were sampled at a vertical resolution of 2 cm using the Denothmeter &#40;2022-2023&#41; and the New Capacitive Sensor from FPGA &#40;2023-2024&#41;. In conditions of ripe snow, dielectric measurements were backed up by melting calorimetry measurements following a recently revised field protocol;<br>   - Snow &#42;&#42;surface roughness&#42;&#42; is expressed by the Root Mean Square of the Heights &#40;RMSH&#41; and the correlation length &#40;CL&#41;, both computed from a digitized snow transect obtained by a digital photograph of a panel vertically inserted into the snow. <br>- &#42;&#42;Snow water equivalent&#42;&#42; was sampled in sections from the surface to the bottom of the snowpack with a cylinder cutter. <br>&amp;nbsp;<br>Additionally, &#42;&#42;runoff&#42;&#42; was automatically measured in the close proximity to the measurement field at a sub-hourly resolution by a lysimeter. The instrument was unfortunately discovered to be clogged at the start of the runoff in 2023 and only repaired in late May 2023. Therefore, the time series is not representative of the runoff start in 2023. The instrument was inspected promptly and assessed as fully functional for the upcoming snow season.
```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
