## Description

Ce jeu de données présente les chiffres mensuels du commerce international du canton de Fribourg et de la Suisse, depuis 2016. Il fournit des informations sur les exportations &#40;EXP&#41; et les importations &#40;IMP&#41; totales, en tonnes et en millions de francs. Les totaux incluent le commerce de l&#x27;or, d&#x27;autres métaux précieux, de pierres précieuses, d&#x27;objets d&#x27;art et d&#x27;antiquités, ainsi que le trafic de perfectionnement à façon, les marchandises en retour et l&#x27;électricité.Les données définitives proviennent de la Statistique du commerce extérieur de l&#x27;Office fédéral de la douane et de la sécurité des frontières &#40;OFDF&#41;. Les données provisoires sont calculées par le Service de la statistique et de la donnée &#40;SSD&#41; du canton de Fribourg à partir de chiffres provisoires et non-plausibilisés de l&#x27;OFDF; elles sont donc susceptibles d&#x27;être révisées lors de la prochaine publication mensuelle, et ce jusqu&#x27;à la publication des données définitives par l&#x27;OFDF, en règle générale en mai de l&#x27;année suivante.Note : Les données du commerce extérieur par canton publiées par l&#x27;OFDF pour l&#x27;année 2025 sont encore provisoires à l&#x27;heure actuelle. Des révisions importantes étant attendues, notamment pour le canton de Fribourg, le SSD a décidé de suspendre la publication des données mensuelles provisoires jusqu&#x27;à la publication des données définitives 2025. Les données mensuelles provisoires 2025 déjà publiées dans ce jeu de données ont été supprimées en date du 10 novembre 2026.La ventilation des données par canton s&#x27;effectue en fonction de l’adresse de l’entreprise exportatrice, respectivement importatrice, propriétaire de la marchandise échangée.Pour plus d&#x27;informations : &#91;https://www.bazg.admin.ch/bazg/fr/home/themes/statistique-du-commerce-exterieur-suisse/faq-aussenhandelsstatistik.html&#93;&#40;https://www.bazg.admin.ch/bazg/fr/home/themes/statistique-du-commerce-exterieur-suisse/faq-aussenhandelsstatistik.html&#41;

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                # Detect file compression after HTTP compression has been decoded.
                if kwargs.get("compression", "infer") == "infer" and source.peek(2).startswith(
                    b"\x1f\x8b"
                ):
                    kwargs["compression"] = "gzip"
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```

```python
# Column types, completeness, and distinct non-missing values.
profile = pd.DataFrame(
    {
        "dtype": df.dtypes.astype(str),
        "missing": df.isna().sum(),
        "missing_pct": df.isna().mean().mul(100),
        "unique": df.nunique(dropna=True),
    }
).sort_values("missing_pct", ascending=False)
display(profile.round(1))
```
