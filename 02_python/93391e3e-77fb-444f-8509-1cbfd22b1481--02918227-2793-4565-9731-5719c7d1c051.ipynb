## Description

Ce jeu de données présente le<br>nombre de travailleurs frontaliers étrangers &#40;titulaires d&#x27;un permis G&#41; actifs<br>occupés dans le canton de Fribourg, ventilé par région ou pays de résidence,<br>par trimestre depuis le premier trimestre 2007.Les données proviennent de la<br>statistique des frontaliers &#40;STAF&#41; de l’Office fédéral de la statistique &#40;OFS&#41;,<br>une statistique de synthèse qui combine les données du Système d’information<br>central sur la migration &#40;SYMIC&#41; et les données des comptes individuels de<br>l’AVS. Elle recense les personnes de nationalité étrangère titulaires d’une<br>autorisation frontalière qui exercent effectivement une activité lucrative en<br>Suisse, et non le simple nombre d’autorisations délivrées. Le permis G<br>&#40;autorisation frontalière&#41; est délivré aux personnes travaillant en Suisse tout<br>en conservant leur domicile principal à l’étranger, où elles retournent<br>régulièrement &#40;au moins une fois par semaine&#41;.Le lieu de travail est le canton<br>de Fribourg dans son ensemble. La ventilation par région ou pays de résidence repose<br>sur une maille hybride : un détail régional pour la France, qui concentre la<br>grande majorité des frontaliers travaillant dans le canton, et une présentation<br>au niveau national pour les autres pays de résidence.Les effectifs ne sont pas des<br>nombres entiers. La méthodologie STAF attribue une pondération fractionnaire<br>&#40;entre 0 et 1&#41; à certains groupes dont le statut d&#x27;activité ne peut pas être<br>établi avec certitude à partir des données AVS : les frontaliers de moins de 18<br>ans, ceux ayant atteint l&#x27;âge de la retraite AVS et les indépendants. Un<br>effectif de 4,09 signifie donc une estimation de l&#x27;ordre de 4 personnes.Données provisoires et révisions<br>: les données des trimestres récents &#40;environ deux ans&#41; sont provisoires et<br>révisées chaque trimestre.<br>Pour plus d&#x27;informations:&#91;https://www.bfs.admin.ch/bfs/fr/home/statistiques/travail-remuneration/activite-professionnelle-temps-travail/population-active/frontaliers.html&#93;&#40;https://www.bfs.admin.ch/bfs/fr/home/statistiques/travail-remuneration/activite-professionnelle-temps-travail/population-active/frontaliers.html&#41;

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
