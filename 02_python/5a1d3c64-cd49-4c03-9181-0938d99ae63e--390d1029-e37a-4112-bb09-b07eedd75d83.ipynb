## Description

Cette donnée représente les zones différenciées des modes de transports selon la Loi pour une mobilité cohérente et équilibrée &#40;LMCE&#41;. Les explications ci-dessous sont extraites de la loi. Par priorisation différenciée, il faut entendre que l&#x27;usage de la voirie disponible est attribué en premier lieu à certains modes de transport notamment au moyen de voies dédiées et par la mise en place d&#x27;un fonctionnement adaptatif ou coordonné des carrefours favorisant les modes de transport désignés. En zones I et II, la priorité en matière de gestion du trafic et d&#x27;aménagement des réseaux est donnée à la mobilité douce et aux transports publics. En zone I : a&#41; le département prend les mesures visant à limiter la vitesse à 30 km/h au maximum selon les conditions prescrites par le droit fédéral; b&#41; l&#x27;accès à cette zone par les autres modes de transport est restreint; c&#41; certains axes à plusieurs voies seront équipés d&#x27;une limitation de vitesse adaptative par des panneaux lumineux. Ces axes seront limités à 30 km/h aux heures de forte demande et à 50 km/h le reste du temps; d&#41; les zones piétonnes ou à priorité piétonne, ainsi que les zones à trafic limité, sont favorisées; e&#41; des dérogations au principe de compensation des places de stationnement énoncé à l&#x27;article 7B de la loi d&#x27;application de la législation fédérale sur la circulation routière, du 18 décembre 1987, tant pour les aménagements améliorant la fluidité que pour ceux agissant sur la sécurité des différents modes de déplacement sont possibles jusqu&#x27;à un maximum de 20%; f&#41; le stationnement des voitures automobiles sur l&#x27;espace public est payant. En zone II : a&#41; le trafic de transit est fortement restreint; b&#41; des axes routiers structurants sont aménagés de facon à garantir la fluidité du transport individuel motorisé; c&#41; des axes sont aménagés de facon à favoriser la circulation des vélos; d&#41; la création de zones 30 est favorisée, selon les conditions prescrites par le droit fédéral et la loi sur les zones 30 et les zones de rencontre, du 21 septembre 2007; e&#41; des dérogations au principe de compensation des places de stationnement énoncé à l&#x27;article 7B de la loi d&#x27;application de la législation fédérale sur la circulation routière, du 18 décembre 1987, tant pour les aménagements améliorant la fluidité que pour ceux agissant sur la sécurité des différents modes de déplacement sont possibles jusqu&#x27;à un maximum de 10%.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
