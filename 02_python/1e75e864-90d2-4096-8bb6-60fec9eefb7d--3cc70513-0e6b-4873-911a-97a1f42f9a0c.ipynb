## Description

Aerosol properties were measured between February 8 and March 31 2019  at the measurement site Weissfluhjoch &#40;LON: 9.806475, LAT: 46.832964&#41;. Optical and aerodynamic particle counters, as well as a scanning mobility particle size spectrometer and an ice nuclei counter were deployed to report particle concentrations and size distributions in fine &#40;10-1000 nm&#41; and coarse mode &#40;- 1000 nm&#41;, cloud condensation nuclei concentrations &#40;CCNCs&#41;, and ice nuclei particle concentrations &#40;ICNCs&#41;. The ambient particles were transported via a heated inlet to be distributed to the particle detecting devices inside the setup room. <br>Optical Particle Counter &#40;OPC&#41;: Light scattering of a diode laser beam caused by travelling particles is used in the both, the OPC-N3 &#40;0.41 - 38.5 &amp;956;m&#41; and GT-526S &#40;0.3 – 5 &amp;956;m&#41;, to determine their size and number concentration. For the OPC-N3, particle size spectra and concentration data are used afterwards to calculate PM&amp;8321;, PM&amp;8322;,&amp;8325; and PM&amp;8321;&amp;8320; &#40;assumptions: particle density: 1.65 g cm&amp;713;&amp;179;, refractive index: 1.5+i0&#41;. <br>Aerodynamic Particle Sizer &#40;APS&#41;: The APS &#40;3321, TSI Inc.&#41; measured the particle size distribution for aerodynamic diameters between 0.5 &amp;956;m and &#126;20 &amp;956;m by the particle’s time-of-flight and light-scattering intensity &#40;assumptions: particle density 1 g cm&amp;713;&amp;179;&#41;. <br>Scanning Mobility Particle Size Spectrometer &#40;SMPS&#41;: Particle number concentrations in a size range between 12 and 460 nm &#40;electrical mobility diameter&#41; were measured at Davos Wolfgang, using a Scanning Mobility Particle Sizer Spectrometer &#40;SMPS 3938, TSI Inc.&#41;. The classifier &#40;3082, TSI Inc.&#41; was equipped with a neutralizer &#40;3088, TSI Inc.&#41; and a differential mobility analyzer working with negative polarity &#40;3081, TSI Inc.&#41;. The size selected particles were counted by a water-based condensation particle counter &#40;3787 TSI Inc.&#41;. The TSI AIM software was used to provide particle size distributions by applying multiple charge and diffusion loss corrections &#40;assumptions: particle density 1 g cm&amp;713;&amp;179;&#41;. <br>Coriolis &amp;956; and LINDA: A microbial air sampler &#40;Coriolis &amp;956;, bertin Instruments&#41; was used to collect airborne particles for investigating their ice nucleating ability with a droplet freezing device. Particles larger than 0.5 &amp;956;m were drawn with an air flow rate of up to 300 l min&amp;8254;&amp;185; into the cone and centrifuged into the wall of the cone due to the forming vortex. The liquid sample was transferred into the LED based Ice Nucleation Detection Apparatus &#40;LINDA, University of Basel&#41; to study heterogeneous ice formation &#40;immersion freezing mode&#41; of ambient airborne particles.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                # Detect file compression after HTTP compression has been decoded.
                if kwargs.get("compression", "infer") == "infer" and source.peek(2).startswith(
                    b"\x1f\x8b"
                ):
                    kwargs["compression"] = "gzip"
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
