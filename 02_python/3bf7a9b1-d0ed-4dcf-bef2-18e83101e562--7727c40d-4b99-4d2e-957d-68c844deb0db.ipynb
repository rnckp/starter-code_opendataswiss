## Description

Decomposition of plant litter is a key process for the transfer of carbon and nutrients in ecosystems. Carbon contained in the decaying biomass is released to the atmosphere as respired CO2, and may contribute to global warming. Litterbag studies have been used to improve our knowledge of the drivers of litter decomposition, but they lack comparability because litter quality is plant species-specific. The use of commercial tea bags as a standard substrate was suggested in order to harmonize studies, where green tea and rooibos represent more labile and more recalcitrant C compounds as surrogates of local litter.<br>The tea bag approach was implemented on eight sites of the Swiss long-term Forest Ecosystem Research &#40;LWF&#41; network &#40;https://www.wsl.ch/LWF&#41;. This allowed us to take advantage from the existing infrastructure and data from a previous litterbag study with local litter. In Beatenberg and Schaenis, additional elevation transects were established &#40;1200-1800 m and 540-1150 m, respectively&#41; to examine particularly the effect of temperature on decomposition. In Pfynwald &#40;https://www.wsl.ch/de/ueber-die-wsl/versuchsanlagen-und-labors/flaechen-im-wald/pfynwald.html&#41; and Salgesch, infrastructure of running projects was used to examine the effect of drought and understory removal, respectively. In Novaggio, tea bags were incubated in summer and winter to study the effect of seasonality particularly precipitation. Tea bags are collected after 3, 12, 24, and 36 months; for the two time-shifted experiments additionally after 6 and 9 months.<br>The study has two primary objectives. Firstly, it contributes to TeaComposition initiative &#40;http://teacomposition.org/&#41; which aims at investigating long-term litter decomposition and its key drivers at present as well as under different future climate scenarios using a common protocol and standard litter &#40;tea&#41; across nine terrestrial biomes. Secondly, the data are used to further develop decomposition models such as Yasso &#40;http://en.ilmatieteenlaitos.fi/yasso&#41; which is used by several countries, including Switzerland to estimate the annual carbon fluxes in dead wood, litter, and soil for reporting in National Greenhouse Gas Inventories under the United Nations Framework Convention on Climate Change and the Kyoto Protocol.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```

```python
# Column types, completeness, and distinct non-missing values.
profile = pd.DataFrame(
    {
        "dtype": df.dtypes.astype(str),
        "missing": df.isna().sum(),
        "missing_pct": df.isna().mean().mul(100),
        "unique": df.nunique(dropna=True),
    }
).sort_values("missing_pct", ascending=False)
display(profile.round(1))
```

```python
# Show up to 20 columns with missing values; percentages use all rows.
missing = profile.loc[profile["missing"] > 0].head(20).rename_axis("column").reset_index()
if not missing.empty:
    fig, ax = plt.subplots(figsize=(8, max(2, len(missing) * 0.3)))
    sns.barplot(data=missing, x="missing_pct", y="column", color="steelblue", errorbar=None, ax=ax)
    ax.set(xlim=(0, 100), xlabel="Missing values (%)", ylabel="", title="Missingness by column")
    fig.tight_layout()
    plt.show()
else:
    print("No missing values to plot (or the dataset has no rows).")
```
