## Description

L&#x27;indice d&#x27;écopotentialité urbaine permet d&#x27;évaluer le degré potentiel de biodiversité d&#x27;un territoire. Il est constitué de 8 indicateurs et mesure la capacité d&#x27;accueil en biodiversité d&#x27;un milieu urbain en fonction de la présence de structures - naturelles ou aménagées - favorables à la biodiversité. Dans le milieu urbain, la biodiversité est contrainte par l&#x27;influence humaine dans un contexte de mutation territoriale. Il existe donc un enjeu fort de conservation et d&#x27;aménagement des espaces verts en ville, afin de lutter contre la perte et la banalisation des espèces, et de conserver la variété, la taille et la connectivité des espaces semi-naturels urbains. Dans ce contexte, les Conservatoire et Jardin Botaniques de la Ville de Genève &#40;CJB&#41; collaborent depuis 2014 avec l&#x27;Office cantonal de l&#x27;agriculture et de la nature &#40;OCAN&#41; dans le but de proposer des outils cartographiques utiles à la prise en considération de cette ressource au travers des planifications directrices et localisées. L&#x27;écopotentialité est un lot d&#x27;indicateurs synthétique de l&#x27;état de la biodiversité pour un périmètre donnée &#40;commune, quartier, zone d&#x27;affectation&#41;. Il a été réfléchi afin de mettre en avant les points forts et les points faibles de la zone d&#x27;étude et donc de faire un lien intéressant avec l&#x27;aménagement futur. L&#x27;indice est construit sur la base de 8 indicateurs normalisés, c&#x27;est-à-dire que chaque plage de valeur des indicateurs est ramenée à une plage normée entre 0 et 1 de manière à permettre une représentation agrégée des 8 indicateurs: 1&#41; Densité de toitures végétalisées &#40;m2/m2&#41; : évaluation de la proportion de surface de toiture végétalisée par rapport aux surfaces bâties d&#x27;un périmètre donné. 2&#41; Surface perméable &#40;%&#41; : évaluation du pourcentage de surface de milieux perméables par rapport aux surfaces imperméables 3&#41; Diversité des milieux verts : regroupement des milieux en 8 catégories assurant des fonctions différentes pour la biodiversité, puis calcul d&#x27;un indice de diversité en milieux verts. 4&#41; Densité d&#x27;arbres &#40;nb/ha&#41;: Nombre d&#x27;arbre issu de l&#x27;identification via des données LIDAR auxquels s&#x27;ajoutent les surfaces forestières &#40;densité moyenne de 370 arbres/ha&#41;. 5&#41; Fragmentation: Mesure la fragmentation du paysage par la présence de barrières physiques &#40;routes, zones bâties, etc.&#41; pour le déplacement de la faune et de la flore entre deux points. 6&#41; Naturalité : Indice compris entre +/- l&#x27;infini, qui distingue les milieux qui ont conservé leurs qualités naturelles &#40;valeurs positives élevées&#41;, des milieux affecté par les activités humaines &#40;valeurs négatives élevées&#41;. 7&#41; Diversité de milieux aquatiques et humides: regroupement des milieux aquatiques en 8 catégories assurant des fonctions différentes pour la biodiversité, puis calcul d&#x27;un indice de diversité en milieux aquatiques. 8&#41; Surface milieux aquatiques: évaluation de la surface de milieu aquatique et humide occupés par rapport à la surface de la zone d&#x27;intérêt. Les données de base utilisées lors des analyses sont notamment la carte des milieux naturels et la couche toitures végétalisées &#40;SIPV&#95;MN&#95;CARTO&#95;5, SIPV&#95;EV&#95;TOIT&#95;VERT&#41;. L&#x27;indice permet de faire un lien intéressant avec l&#x27;aménagement et la gestion du territoire en identifiant les facteurs manquants ou favorables à la biodiversité sur le périmètre considéré. Les périmètres d&#x27;étude vont de l&#x27;échelle de la commune, jusqu&#x27;à la zone d&#x27;aménagement &#40;1500m2&#41;. C&#x27;est pourquoi, cet indice a été intégré en annexe du guide de densification de la zone 5 &#40;zone villa&#41; publié par l&#x27;Office de l&#x27;urbanisme en 2017. Il est par ailleurs utilisé par les communes et leurs mandataires dans le cadre de la révision des plans directeurs communaux et des plans guides des Grands projets.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
