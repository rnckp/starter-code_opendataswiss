## Description

Study Aim<br>We collected these data to alternatively train and validate high resolution &#40;&#126; 90 m&#41; Species Distribution Models &#40;SDMs&#41; and Species Abundance Models &#40;SAMs&#41; for &#95;Betula nana&#95; L. &#40;dwarf birch, Betulaceae&#41; and &#95;Salix glauca&#95; L. &#40;grey willow, Salicaceae&#41; in Southwest Greenland to assess how well such models can predict local-scale patterns.<br> Data Description<br>Individual &#40;presence-absence, abundance, maximum vegetative height&#41; and community &#40;species composition, maximum canopy height&#41; shrub data for two fjords near Nuuk, Southwest Greenland. Also provided are corresponding downscaled climate data as well as calculated topographic and terrain wetness indicator variables. <br> Nuup Kangerlua &#40;Godthåbsfjord&#41;<br>&#95;Betula nana&#95; and &#95;Salix glauca&#95; presence-absence, abundance, community species richness<br> Kangerluarsunnguaq &#40;Kobbefjord&#41;<br>Shrub presence-absence, abundance, maximum vegetative height, community composition, maximum shrub canopy height <br> Methods<br> Field survey in Nuup Kangerlua<br>We conducted a stratified systematic plant survey along the length of Nuup Kangerlua &#40;NK&#41; fjord in Soutwesth Greenland &#40;Fig. 1 in Chardon et al. 2022; following Nabe-Nielsen et al., 2017&#41;. At five distinct sites, we sampled along elevational gradients to collect data on presences, absences, abundance, and species composition of all woody species using a 0.7 x 0.7 m pin-point frame &#40;Fig. 1e in Chardon et al. 2022&#41;. For model training, we converted these pin-point data to percent cover estimates based on the number of pins dropped &#40;n = 25 per plot&#41; and averaged them across the 119 spatio-climatic grids &#40;see next section&#41; corresponding to the plot locations &#40;for details see Appendix S2 in Chardon et al. 2022&#41;. <br> Field survey in Kangerluarsunnguaq <br>We conducted a random stratified plant survey in Kangerluarsunnguaq &#40;K&#41; fjord in Southwest Greenland. We used a preliminary Species Abundance Model trained with summed pin counts of &#95;Betula nana&#95; in NK fjord &#40;see Fig. S1.3 in Chardon et al. 2022&#41; to stratify the &#126; 27 x 17 km fjord landscape into low, medium, and high abundances classes. We randomly selected 90 x 90 m spatio-climatic grids to survey in each class for a total of 200 grids, ensuring that they were accessible by foot or boat &#40;for details see Appendix S2 in Chardon et al. 2022&#41;. Within each grid, we sampled within three 1 m2 quadrats arranged in a randomly rotated equilateral triangle centered on the mid-point of the cell. We used a gridded sampling quadrat with 1% delineations &#40;Fig. 1h in Chardon et al. 2022&#41; to record woody species presences, absences, and composition, estimated percent cover, and measured maximum shrub species vegetatitve height. At every plot, we also visually scanned the area in a 20 m radius from the plot and recorded the presence of any additional shrub species to estimate grid-level species richness. As in NK fjord, we averaged these data at the grid level &#40;for details see Appendix S2 in Chardon et al. 2022&#41;.<br> Biotic variables<br>We calculated biotic microscale variables from the plant survey data collected in NK and K fjords. We calculated shrub species richness, diversity, and competition &#40;i.e. sum of non-B. nana or non-S. glauca pin hits or percent cover&#41;. In K fjord, we also calculated canopy height as the community weighted mean &#40;by abundance&#41; of maximum vegetative shrub height.<br> Climate variables <br>We computed high resolution temperature, precipitation, and insolation for local scale data for the study area by statistically downscaling climate time series &#40;1982 - 2013&#41; from the monthly CHELSA data &#40;Karger et al. 2017&#41;. We downscaled these data from 30 arc sec &#40;&#126; 400 m at the latitude of our study&#41; to our target grid size of &#126; 90 m with geographic weighted regression and using the MEaSUREs Greenland Ice Mapping Project &#40;GIMP&#41; Digital Elevation Model &#40;DEM&#41; v. 1 &#40;Howat et al., 2014, 2015&#41;. We then calculated 30-year averages of the climate parameters: average summer &#40;June – August&#41; maximum temperature, yearly maximum temperature, yearly minimum temperature, temperature continentality &#40;yearly max. - min. temperatures&#41;, cumulative Spring &#40;March – May&#41; precipitation, cumulative summer precipitation, and average summer incident solar radiation &#40;henceforth, insolation&#41; &#40;for calculation details see Appendices S2, S3 in Chardon et al. 2022 and Appendix S2 in von Oppen et al. 2021&#41;.<br> Topography and terrain wetness indicator variables<br>We calculated several topographic and terrain wetness indices at a local scale. We derived slope, aspect, and the SAGA wetness index &#40;hereafter TWI; Boehner et al., 2002; Boehner and Selige, 2006&#41; from the GIMP DEM. TWI is a measure of how ‘wet’ an area is, based on water drainage from the surrounding landscape. We also calculated the tasseled cap wetness component &#40;hereafter TCW, Crist and Cicone 1984&#41; from satellite images &#40;for details see Appendices S2, S3 in Chardon et al. 2022&#41; as an alternative measure of wetness. <br> Computer code<br>Attached as zip file and available on GitLab &#40;https://gitlab.com/nathaliechardon/gl&#95;microclim&#41;<br> Third-party data<br>Data used to calculate climate, topography, and terrain wetness indicator variables are publicly available &#40;see Appendix S2 in Chardon et al. 2022 for all data references&#41;.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                # Detect file compression after HTTP compression has been decoded.
                if kwargs.get("compression", "infer") == "infer" and source.peek(2).startswith(
                    b"\x1f\x8b"
                ):
                    kwargs["compression"] = "gzip"
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
