## Description

OTEMO est l&#x27;acronyme de l&#x27;Ordonnance technique du DDPS sur la mensuration officielle du 10 juin 1994. Cette ordonnance fédérale introduit un catalogue d&#x27;objets de couverture du sol cantonale &#40;art.7-b&#41;. Selon cette ordonnance, la couche d&#x27;information &quot;couverture du sol&quot; couvre l&#x27;entier du territoire et distingue 6 genres &#40;renseignés sous CODE&#95;OTEMO&#95;1&#41; et 25 sous-genres &#40;renseignés sous CODE&#95;OTEMO&#95;2&#41; d&#x27;objets: 1. Bâtiments ; 2. Surfaces à revêtement dur, subdivisées en: route/chemin, trottoir, îlot, chemin de fer, place d&#x27;aviation, bassin et autres surfaces à revêtement dur ; 3. Surfaces vertes, subdivisées en: champ/pré/pâturage, culture intensive &#40;elle-même subdivisée en vigne et autre culture intensive&#41;, jardin, haut- et bas-marais et autre surface verte ; 4. Eaux, subdivisées en: eau stagnante, cours d&#x27;eau et roselière ; 5. Surfaces boisées, subdivisées en: forêt dense, pâturage boisé &#40;lui-même subdivisé en «pâturage boisé dense» et «pâturage boisé ouvert»&#41; et autre surface boisée ; 6. Surfaces sans végétation, subdivisées en: rocher, glacier/névé, éboulis/sable, extraction de matériaux/décharge et autre surface sans végétation. Afin de pouvoir répondre aux exigences fédérales, la carte OTEMO compile les données &quot;métier&quot; issues de: - la Mensuration Officielle &#40;MO&#41;, sous la responsabilité de la Direction de l&#x27;information du territoire &#40;DIT&#41; ; - la carte des milieux naturels du Grand Genève &#40;MN&#41;, diffusée par le consortium Système d&#x27;information du patrimoine vert &#40;SIPV&#41; sous la responsabilité du Jardin Botanique de Genève. Les données cadastrales de la MO définissent la propriété foncière et servent de référence au système d&#x27;information du territoire à Genève &#40;SITG&#41;. Elles servent également à l&#x27;établissement du plan du registre foncier, des états descriptifs des parcelles, du plan de base et du plan de ville. Ces données se fondent sur le modèle de données fédéral MD.01-MO-CH, qui organise le contenu en couches d&#x27;informations thématiques telles que la couverture du sol. La carte MN vise à inventorier les milieux naturels du canton de Genève à partir de différentes sources de données telles que les orthophotos, les données cadastrales, des observations terrain, et autres contributions tierces. La carte est mise à jour chaque année avec les données des CJBG, les couches de la MO et les données acquises par des tiers &#40;Office cantonal de l&#x27;agriculture et de la nature, communes, bureaux, etc.&#41;. A l&#x27;heure actuelle, le canton n&#x27;est pas prospecté de manière uniforme: les zones publiques sont davantage prospectées que les zones privées ou non-accessibles. Lorsque des surfaces sont relevées par des géomètres, elles sont rapportées dans les couches de la MO puis remontées dans la carte MN, selon une table d&#x27;équivalence mise en place par le CJGB et la DIT. En retour, la carte MN est utilisée pour documenter les surfaces du canton qui ne sont pas relevées dans les données cadastrales officielles fournies par la MO. Un script FME est utilisé à la DIT pour convertir les données d&#x27;un modèle à l&#x27;autre et automatiser les flux. En sortie, la DIT fournit une carte en partition complète du territoire cantonal selon la norme OTEMO.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
