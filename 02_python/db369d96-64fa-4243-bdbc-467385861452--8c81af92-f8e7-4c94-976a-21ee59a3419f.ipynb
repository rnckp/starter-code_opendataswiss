## Description

The dataset comprises - 90 000 records from inventories in 54 strict forest reserves in &#91;Switzerland&#93;&#40;https://www.wsl.ch/de/wald/biodiversitaet-naturschutz-urwald/naturwaldreservate.html&#41; and &#91;Lower Saxony / Germany&#93;&#40;http://naturwaelder.de/&#41; along a considerable environmental gradient. It was used to develop parsimonious, species-specific mortality models for 18 European tree species based on tree size and growth as well as additional covariates on stand structure and climate. <br> Inventory data<br>Measurements had been conducted repeatedly on up to 14 permanent plots per reserve for up to 60 years with re-measurement intervals of 4 - 27 years. The permanent plots vary in size between 0.03 and 3.47 ha. The inventories provide diameter measurements at breast height &#40;DBH&#41; and information on the species and status &#40;alive or dead&#41; of trees with DBH ≥ 4 cm for Switzerland and ≥ 7 cm for Germany. <br> Data selection<br>We excluded three permanent plots where at least 80 % of the trees died during an interval of 10 years, and mortality could be clearly assigned to a disturbance agent. Mortality in the remaining stands was rather low, with a mean annual mortality rate of 1.5 % and strong variation between plots from 0 to 6.5 % &#40;assessed for trees of all species with DBH ≥ 7 cm&#41;.<br>We only used data from permanent plots with at least 20 trees per species to obtain reliable plot-level mortality rates even for species with low mortality rates &#40;about 5 % during 10 years&#41;, and selected tree species occurring on at least 10 plots to cover sufficient ecological gradients. This led to a dataset of 197 permanent plots and 18 tree or shrub species: &#95;Abies alba&#95; Mill., &#95;Acer campestre&#95; L., &#95;Acer pseudoplatanus&#95; L., &#95;Alnus incana&#95; Moench., &#95;Betula pendula&#95; Roth, &#95;Carpinus betulus&#95; L., &#95;Cornus mas&#95; L., &#95;Corylus avellana&#95; L., &#95;Fagus sylvatica&#95; L., &#95;Fraxinus excelsior&#95; L., &#95;Picea abies&#95; &#40;L.&#41; Karst, &#95;Pinus mugo&#95; Turra, &#95;Pinus sylvestris&#95; L., &#95;Quercus pubescens&#95; Willd., &#95;Quercus&#95; spp. &#40;&#95;Q. petraea&#95; Liebl. and &#95;Q. robur&#95; L.; not properly differentiated in the Swiss inventories&#41;, &#95;Sorbus aria&#95; Crantz, &#95;Tilia cordata&#95; Mill. and &#95;Ulmus glabra&#95; Huds.. <br> Predictors of tree mortality<br>We considered tree size and growth as key indicators for mortality risk. Radial stem growth between the first and second inventory and DBH at the second inventory were used to predict tree status &#40;alive or dead&#41; at the third inventory. To this end, the annual relative basal area increment &#40;relBAI&#41; was calculated as the compound annual growth rate of the trees basal area. Additional covariates on stand structure and climate comprise mean annual precipitation sum &#40;P&#41;, mean annual air temperature &#40;mT&#41;, the mean and the interquartile range of DBH &#40;mDBH, iqrDBH&#41;, basal area &#40;BA&#41; and the number of trees &#40;N&#41; per hectare. <br> Further information<br>For further information, refer to Hülsmann &#95;et al&#95;. &#40;in press&#41; How to kill a tree – Empirical mortality models for eighteen species and their performance in a dynamic forest model. &#95;Ecological Applications&#95;.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                # Detect file compression after HTTP compression has been decoded.
                if kwargs.get("compression", "infer") == "infer" and source.peek(2).startswith(
                    b"\x1f\x8b"
                ):
                    kwargs["compression"] = "gzip"
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
