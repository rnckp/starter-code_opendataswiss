## Description

Le référentiel des projets urbains &#40;RefPU&#41; est la version numérique des plans d&#x27;affectations produits par l&#x27;autorité compétente en matière d&#x27;aménagement du territoire du canton de Genève. Pour l&#x27;heure, seuls les plans localisés de quartier &#40;PLQ&#41; sont publiés dans ce référentiel. A terme, les modifications de zone &#40;MZ&#41; ainsi que les plans directeurs de zones industrielles &#40;PDZI&#41;, mais aussi les plans de site &#40;PS&#41; produits par l&#x27;office du patrimoine et des sites, pourraient également y être publiés. La thématique RefPU permet de consulter l&#x27;ensemble des données géographiques et alphanumériques contenues dans un PLQ et de les mettre en regard avec d&#x27;autres données territoriales issues d&#x27;autres politiques publiques ou des données de référence &#40;limites communales, cadastre, courbes de niveau, etc.&#41;. Les PLQ publiés sur ce thème font l&#x27;objet d&#x27;une consultation publique, conformément aux exigences légales en vigueur. Les données qu&#x27;ils contiennent sont mises à la disposition des partenaires du SITG et consultables sur ce guichet. Néanmoins, hormis le périmètre de validité d&#x27;un plan adopté qui a une valeur légale contraignante, les autres données géographiques et alphanumériques mises à disposition, ne créent aucun autre droit ou obligation que ceux découlant des textes et documents légalement mis en procédure, adoptés et publiés. En premier lieu, le RefPU permet de géoréférencer chaque plan, et par extension, chacun des éléments constitutifs du plan. Il offre également une unicité de représentation graphique standardisée pour l&#x27;ensemble des plans localisés de quartier grâce à une charte graphique et un modèle de données géomatique uniformisés. La thématique RefPU offre également la possibilité de représenter simultanément dans une carte numérique l&#x27;ensemble des PLQ numérisés sur tout le territoire genevois. Grand avantage par rapport au format papier ou pdf, le guichet numérique permet d&#x27;interroger la valeur de chaque donnée contenue dans une couche du modèle. Note: tout comme sur un plan en format papier ou pdf, les PLQ numériques se composent d&#x27;un plan d&#x27;aménagement et d&#x27;un plan des équipements, distincts. Les deux plans peuvent être consultés séparément.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                # Detect file compression after HTTP compression has been decoded.
                if kwargs.get("compression", "infer") == "infer" and source.peek(2).startswith(
                    b"\x1f\x8b"
                ):
                    kwargs["compression"] = "gzip"
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
