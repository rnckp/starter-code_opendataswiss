# **Stumps as a dead wood resource in forests - data based on the Swiss National Forest Inventory**


## Description

Based on the detailed tree stump inventory implemented in the Swiss NFI5 &#40;https://www.lfi.ch/lfi/lfi.php&#41;, a study was conducted to obtain an accurate assessment of the stumps pool in the Swiss NFI over the last 30 years and to identify its significance for the total dead wood &#40;DW&#41; pool. The NFI5 includes a detailed stump inventory to improve accuracy and completeness of the above-ground DW- pool. Based on available data, stump volume estimates were derived at different accuracies to evaluate the contribution to the total DW-pool over time.<br>The study found that in Swiss Forests the contribution of stumps to total DW-pool is approximately 25%, and that applying simplifying assumptions to estimate stump volume can result in a significant underestimation of the true size of this pool. This study demonstrates that stumps can be a significant proportion of DW in forests, which should be accounted for in order to improve accuracy and completeness of NFI estimates and derived data such as C stocks for greenhouse gas reporting.<br>The study is published in Annals of Forest Science &#40;2022&#41; 79:34, https://doi.org/10.1186/s13595-022-01155-7 &#40;open access&#41;.<br>The data can be obtained from the authors upon reasonable request.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
