## Description

Droit distinct et permanent &#40;DDP&#41; : Immeuble selon l&#x27;art. 943 al. 1 du Code Civil Suisse. L&#x27;immatriculation et la description de chaque immeuble dans le registre foncier s&#x27;effectuent sur la base de la mensuration officielle, notamment d&#x27;un plan du registre foncier &#40;CC art. 950 al.1&#41;. La loi fédérale du 5 octobre 2007 sur la géoinformation &#40;LGéo&#41; fixe les exigences qualitatives et techniques applicables à la mensuration officielle. Il s&#x27;agit avant tout de servitudes de superficie, personnelles et cessibles &#40;droit distinct&#41;, constituées pour une durée indéterminée ou de 30 ans au moins et de 100 ans au plus &#40;droit permanent&#41;, inscrites au registre foncier sur un feuillet qui leur est propre ou qui pourraient l&#x27;être en tout temps à la seule demande du bénéficiaire. Que l&#x27;immatriculation soit requise ou non, le DDP est soumis aux règles propres aux immeubles pour la création et pour la cession. Un DDP est possible pour : - les sources &#40;art 780 CCS&#41; - les superficies &#40;art 675 et 779 CCS&#41; - les plantes isolées et les plantations &#40;art 678 al.2 et 3 CCS&#41; Avantages de l&#x27;immatriculation d&#x27;un DDP : - possibilité de les grever de droits réels limités, notamment des gages; - la cessibilité - elle doit être la plus large possible, toutefois le contrat constitutif peut restreindre le cercle des acquéreurs potentiel Le propriétaire du sol ou superficiant conserve son bien-fonds tout en le mettant à disposition d’un tiers. Il s’en garantit un rendement, en fixant généralement une rente de superficie annuelle. Le superficiaire peut y ériger un bâtiment ou une autre construction sans devoir acquérir le sol, ce qui lui évite une partie importante des coûts d’investissement. A l&#x27;expiration du droit de superficie, les installations font retour au propriétaire du fonds Un droit de préemption légal et réciproque du propriétaire du fonds et du superficiaire est constitué par l&#x27;art.682 al 2 CCS. Il peut être modifié ou supprimé par acte authentique avec annotation au registre foncier.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                # Detect file compression after HTTP compression has been decoded.
                if kwargs.get("compression", "infer") == "infer" and source.peek(2).startswith(
                    b"\x1f\x8b"
                ):
                    kwargs["compression"] = "gzip"
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
