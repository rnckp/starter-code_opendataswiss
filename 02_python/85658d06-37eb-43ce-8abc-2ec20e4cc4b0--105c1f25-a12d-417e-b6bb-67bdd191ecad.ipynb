## Description

Ergänzungsleistungen &#40;EL&#41; werden Personen mit einer AHV- oder IV-Rente ausbezahlt, wenn ihr Einkommen die minimalen Lebenskosten nicht deckt. <br><br>Dieser Datensatz zeigt, wie viele Personen im Kanton Zürich zusätzlich zu ihrer AHV- oder IV-Rente EL beziehen und wie hoch die Bezugsquote ist, also der Anteil der EL-Rentenbeziehenden gemessen an allen AHV- respektive IV-Beziehenden. Da der Anspruch auf EL auf Haushaltsebene berechnet wird, wird neben der Anzahl Rentenbeziehenden auch die Anzahl Rentenbeziehende einschliesslich ihrer Ehepartnerinnen oder Ehepartner im gleichen Haushalt ausgewiesen. Die Anzahl der AHV-/IV-Rentenbeziehenden dient als Referenzbevölkerung für die Berechnung der Bezugsquote und wird ebenfalls ausgewiesen.<br><br>Die Werte liegen nach den Merkmalen Altersklasse, Geschlecht und Nationalität vor – sowohl einzeln als auch in Kombination. Räumlich beziehen sich alle Werte auf den gesamten Kanton Zürich. Eine Aufteilung nach Gemeinden oder Bezirken ist nicht enthalten. Der Datensatz beginnt im Jahr 2010 und wird laufend um neue Jahre ergänzt. <br><br>Die Daten stammen aus dem Ergänzungsleistungen-Register und dem Rentenregister des Bundesamts für Sozialversicherungen &#40;BSV&#41;. <br><br>Weitere Angaben zu Datensatz-Struktur und verwendeten Merkmalen sind in der Variablenbeschreibung zu finden.

```python
from io import BufferedReader
from typing import BinaryIO

import requests

USER_AGENT = "OpendataSwiss-StarterCode-Python/1.0"
HTTP_TIMEOUT = 30


def get_dataset(
    url: str | BinaryIO,
    *,
    sep: str | None = None,
    drop_empty_unnamed: bool = True,
    **kwargs: object,
) -> pd.DataFrame:
    """Read one CSV resource; preserve data and report parsing failures."""
    if isinstance(url, str) and url.startswith(("https://", "http://")):
        with requests.get(
            url, headers={"User-Agent": USER_AGENT}, timeout=HTTP_TIMEOUT, stream=True
        ) as response:
            response.raise_for_status()
            # Decode HTTP compression while streaming CSV bytes into pandas.
            response.raw.decode_content = True
            # Let the buffer finish reading after the HTTP connection reaches EOF.
            response.raw.auto_close = False
            with BufferedReader(response.raw) as source:
                # Detect file compression after HTTP compression has been decoded.
                if kwargs.get("compression", "infer") == "infer" and source.peek(2).startswith(
                    b"\x1f\x8b"
                ):
                    kwargs["compression"] = "gzip"
                return get_dataset(source, sep=sep, drop_empty_unnamed=drop_empty_unnamed, **kwargs)
    data = pd.read_csv(url, sep=sep, engine="python", dtype_backend="numpy_nullable", **kwargs)
    # Extra fields can make pandas silently consume leading values as an index.
    if kwargs.get("index_col") is None and not isinstance(data.index, pd.RangeIndex):
        raise pd.errors.ParserError(
            "CSV rows contain unexpected index fields. Check the delimiter and header; "
            "set index_col explicitly only if these fields are an intentional index."
        )
    if drop_empty_unnamed and len(data) > 0:
        # Blank CSV headers (often from trailing separators) become "Unnamed: n".
        unnamed = data.columns.astype(str).str.fullmatch(r"Unnamed: \d+(?:\.\d+)?")
        empty_unnamed = unnamed & data.isna().all()
        if empty_unnamed.any():
            print("Removed empty unnamed columns:", data.columns[empty_unnamed].tolist())
            data = data.loc[:, ~empty_unnamed]
    return data
```
